### Ajustando formato data e hora

```python
import pandas as pd

# Supondo que você tenha um DataFrame com a coluna 'time' já convertida para datetime
df = pd.DataFrame({'time': ['2019-10-06 18:56:26', '2020-11-15 09:30:45', '2021-08-25 14:20:10']})
df['time'] = pd.to_datetime(df['time'])


df['dia'] = df['time'].dt.day
df['mes'] = df['time'].dt.month
df['ano'] = df['time'].dt.year
df['hora'] = df['time'].dt.hour
df['minuto'] = df['time'].dt.minute
df['segundo'] = df['time'].dt.second

# Exibindo o DataFrame resultante
print(df)
```

### Excluindo todas as linhas com NaN

```python
import pandas as pd
import numpy as np

# Exemplo de DataFrame com valores NaN
df = pd.DataFrame({'A': [1, np.nan, 3, np.nan],
                   'B': [np.nan, 5, np.nan, 7],
                   'C': [9, 10, np.nan, np.nan]})

# Substituindo os NaN por 0 em todo o DataFrame
df_preenchido = df.fillna(0)

# Exibindo o DataFrame resultante
print(df_preenchido)
```

### Gráfico de barras

```python
import matplotlib.pyplot as plt

# Dados fictícios: distritos e número de votos
distritos = ['Distrito A', 'Distrito B', 'Distrito C', 'Distrito D', 'Distrito E']
votos = [5000, 7000, 3000, 9000, 6000]

# Criar o gráfico de barras
plt.figure(figsize=(10, 6))
plt.bar(distritos, votos)
plt.title('Número de Votos por Distrito')
plt.xlabel('Distritos')
plt.ylabel('Número de Votos')
plt.xticks(rotation=45)  # Rotaciona os rótulos do eixo x para facilitar a leitura
plt.grid(axis='y')  # Adiciona grades apenas no eixo y

# Exibir o gráfico
plt.tight_layout()  # Ajusta automaticamente os espaços para evitar sobreposição de elementos
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Vamos criar um DataFrame de exemplo para simular a situação
data = {
    'Localidade': ['Localidade A', 'Localidade B', 'Localidade C', 'Localidade A', 'Localidade B', 'Localidade C'],
    'Partido': ['Partido X', 'Partido X', 'Partido X', 'Partido Y', 'Partido Y', 'Partido Y'],
    'Votos': [100, 150, 120, 80, 90, 100]
}
df = pd.DataFrame(data)

# Definir as localidades e partidos de interesse
localidades_interesse = ['Localidade A', 'Localidade B']
partidos_interesse = ['Partido X', 'Partido Y']

# Filtrar o DataFrame para incluir apenas as localidades e partidos de interesse
df_filtrado = df[(df['Localidade'].isin(localidades_interesse)) & (df['Partido'].isin(partidos_interesse))]

# Agrupar os dados por partido e localidade e somar os votos
dados_agrupados = df_filtrado.groupby(['Localidade', 'Partido'])['Votos'].sum().unstack()

# Plotar o gráfico de barras com as barras lado a lado
plt.figure(figsize=(6, 4))
largura_barra = 0.4  # largura das barras
deslocamento = largura_barra / 2  # deslocamento horizontal para as barras lado a lado
for i, partido in enumerate(partidos_interesse):
    posicoes_x = [x + i * largura_barra - deslocamento for x in range(len(dados_agrupados.index))]
    plt.bar(posicoes_x, dados_agrupados[partido], width=largura_barra, label=partido)

plt.xlabel('Localidades')
plt.ylabel('Votos')
plt.title('Votos por Partido em Localidades de Interesse')
plt.xticks(range(len(dados_agrupados.index)), dados_agrupados.index)  # definir rótulos do eixo x
plt.legend()
plt.tight_layout()

# Exibir o gráfico
plt.show()
```

### Excluindo elementos com critérios de um DataFrame

```python
import pandas as pd

# Exemplo de DataFrame
data = {'date': ['2019-10-06', '2019-10-06', '2019-10-06'],
        'time': ['19:00:02', '20:00:00', '20:15:00'],
        'totalVoters': [1152050, 1152050, 1152050]}
df = pd.DataFrame(data)

# Excluindo linhas com hora inferior a 20:10:05 na coluna 'time'
df = df.drop(df.loc[df['time'] < '20:10:05'].index)

# Exibindo o DataFrame após a exclusão das linhas
print(df)
```
